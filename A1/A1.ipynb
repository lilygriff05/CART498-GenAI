```python
# Verify GPU access and import diffusion model

import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

from diffusers import DiffusionPipeline

pipeline = DiffusionPipeline.from_pretrained(
  "Lykon/dreamshaper-8", dtype=torch.bfloat16, device_map=device,
)
```

**PROMPTS**

```python
prompt = """
A 1990s style film poster with a gigantic cat terrorizing new york city, big city
highly detailed, collage style, cinemascope, noir, epic, paper grain
"""
pipeline(prompt).images[0]
```

```python
prompt = """
A 1990s style film poster with a gigantic cat terrorizing new york city, broken windows, fire
highly detailed, collage style, cinemascope, film noir, epic, paper grain, dynamic
"""
pipeline(prompt).images[0]
```
